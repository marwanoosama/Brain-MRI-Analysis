# Medical Image Analysis Assistant — Results Report

**Dataset:** Brain MRI Images for Brain Tumor Detection  
**Task:** Binary classification (Normal vs Abnormal)  
**Pipeline:** Preprocessing → Segmentation → Feature Extraction → Classification

## Setup

```python
import os
import sys
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, jaccard_score, confusion_matrix, ConfusionMatrixDisplay

# Add project root to path so we can import src modules
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, project_root)

from src.preprocess import preprocess
from src.segment import segment
from src.features import extract_features, extract_texture, extract_shape
from src.classify import train_classifier

print('Setup complete.')
```

## Step 0 — Load Dataset

```python
data_dir = os.path.join(project_root, 'data')
normal_dir = os.path.join(data_dir, 'normal')
abnormal_dir = os.path.join(data_dir, 'abnormal')

image_paths = []
labels = []

for fname in sorted(os.listdir(normal_dir)):
    fpath = os.path.join(normal_dir, fname)
    if os.path.isfile(fpath):
        image_paths.append(fpath)
        labels.append(0)

for fname in sorted(os.listdir(abnormal_dir)):
    fpath = os.path.join(abnormal_dir, fname)
    if os.path.isfile(fpath):
        image_paths.append(fpath)
        labels.append(1)

print(f'Normal images: {labels.count(0)}')
print(f'Abnormal images: {labels.count(1)}')
print(f'Total: {len(labels)}')
```

## Step 1 — Image Preprocessing

Apply **Gaussian Filter** (reduce high-frequency noise) followed by **Median Filter** (remove salt-and-pepper noise while preserving edges).

```python
# Show preprocessing on 3 sample images (1 normal, 2 abnormal)
sample_indices = [0, labels.count(0), labels.count(0) + 1]

fig, axes = plt.subplots(3, 2, figsize=(10, 12))
for i, idx in enumerate(sample_indices):
    original = cv2.imread(image_paths[idx], cv2.IMREAD_GRAYSCALE)
    preprocessed = preprocess(image_paths[idx])
    label_name = 'Normal' if labels[idx] == 0 else 'Abnormal'

    axes[i, 0].imshow(original, cmap='gray')
    axes[i, 0].set_title(f'Original ({label_name})')
    axes[i, 0].axis('off')

    axes[i, 1].imshow(preprocessed, cmap='gray')
    axes[i, 1].set_title(f'Preprocessed ({label_name})')
    axes[i, 1].axis('off')

plt.tight_layout()
plt.show()
```

## Step 2 — Segmentation

Extract the Region of Interest (ROI) using **Otsu's Thresholding**.

```python
# Show segmentation on the same 3 samples
fig, axes = plt.subplots(3, 3, figsize=(14, 12))
for i, idx in enumerate(sample_indices):
    preprocessed = preprocess(image_paths[idx])
    roi, mask = segment(preprocessed)
    label_name = 'Normal' if labels[idx] == 0 else 'Abnormal'

    axes[i, 0].imshow(preprocessed, cmap='gray')
    axes[i, 0].set_title(f'Preprocessed ({label_name})')
    axes[i, 0].axis('off')

    axes[i, 1].imshow(mask, cmap='gray')
    axes[i, 1].set_title(f'Segmentation Mask')
    axes[i, 1].axis('off')

    axes[i, 2].imshow(roi, cmap='gray')
    axes[i, 2].set_title(f'ROI')
    axes[i, 2].axis('off')

plt.tight_layout()
plt.show()
```

## Step 3 — Feature Extraction

Extract **texture features** (GLCM: contrast, energy, homogeneity, correlation) and **shape features** (area, perimeter, circularity) from each image.

```python
# Process all images through the pipeline
all_features = []
all_masks = []
valid_labels = []

for i, (path, label) in enumerate(zip(image_paths, labels)):
    try:
        img = preprocess(path)
        roi, mask = segment(img)
        feats = extract_features(roi, mask)
        all_features.append(feats)
        all_masks.append(mask)
        valid_labels.append(label)
    except Exception as e:
        print(f'Skipping {os.path.basename(path)}: {e}')

features = np.array(all_features)
labels_arr = np.array(valid_labels)

print(f'Feature matrix shape: {features.shape}')
print(f'Features per image: contrast, energy, homogeneity, correlation, area, perimeter, circularity')
print(f'\nSample feature vector (first image):')
print(f'  {features[0]}')
```

## Step 4 — Classification

Train an **SVM (Support Vector Machine)** with RBF kernel on the extracted features.  
Split: 80% train / 20% test.

```python
clf, scaler, X_test, y_test, y_pred = train_classifier(features, labels_arr)

print(f'Test set size: {len(y_test)}')
print(f'Predictions: {y_pred}')
print(f'Actual:      {y_test}')
```

## Step 5 — Evaluation

Compute all 4 required metrics: **Accuracy**, **Precision/Recall**, **IoU**, and **Matching Accuracy**.

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

# IoU for segmentation masks
from sklearn.model_selection import train_test_split
_, test_idx = train_test_split(range(len(features)), test_size=0.2, random_state=42)
gt_masks_test = [all_masks[i] for i in test_idx]
pred_masks_test = [all_masks[i] for i in test_idx]

iou_scores = []
for gt, pred in zip(gt_masks_test, pred_masks_test):
    gt_flat = (gt.flatten() > 0).astype(int)
    pred_flat = (pred.flatten() > 0).astype(int)
    iou_scores.append(jaccard_score(gt_flat, pred_flat, average='binary'))
iou = np.mean(iou_scores)

matching_accuracy = accuracy

print('=' * 50)
print('EVALUATION RESULTS')
print('=' * 50)
print(f'  Accuracy:           {accuracy:.4f}')
print(f'  Precision:          {precision:.4f}')
print(f'  Recall:             {recall:.4f}')
print(f'  IoU (Jaccard):      {iou:.4f}')
print(f'  Matching Accuracy:  {matching_accuracy:.4f}')
print('=' * 50)
```

### Metrics Summary Table

| Metric | Value | Description |
|---|---|---|
| **Accuracy** | — | % of all images correctly classified |
| **Precision** | — | Trade-off: false positives |
| **Recall** | — | Trade-off: false negatives |
| **IoU (Jaccard)** | — | Overlap between predicted segmentation mask and ground truth |
| **Matching Accuracy** | — | Per-image label match rate |

### Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal', 'Abnormal'])

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(ax=ax, cmap='Blues', values_format='d')
ax.set_title('Confusion Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Full Pipeline Visualization

Sample images at each stage of the pipeline: Original → Preprocessed → Segmentation Mask → ROI

```python
n_samples = 5
fig, axes = plt.subplots(n_samples, 4, figsize=(16, 4 * n_samples))
stage_titles = ['Original', 'Preprocessed', 'Segmentation Mask', 'ROI']

for i in range(n_samples):
    path = image_paths[i]
    original = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    preprocessed = preprocess(path)
    roi, mask = segment(preprocessed)

    images = [original, preprocessed, mask, roi]
    for j, (img, title) in enumerate(zip(images, stage_titles)):
        axes[i, j].imshow(img, cmap='gray')
        if i == 0:
            axes[i, j].set_title(title, fontsize=14)
        axes[i, j].axis('off')

plt.tight_layout()
plt.show()
```
